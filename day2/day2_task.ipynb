Drop useless data


```python
dataset = dataset.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'], axis=1)
dataset.head()
```

handle missing value

```python
# Age
dataset['Age'] = dataset['Age'].fillna(dataset['Age'].median())
dataset.isnull().sum()
```

```python
# Emnarked
dataset['Embarked'] = dataset['Embarked'].fillna(dataset['Embarked'].mode()[0])
dataset.isnull().sum()
```

Filter female and Group by class

```python
female_dataset = dataset[dataset['Sex'] == 'female']
group_female_dataset = female_dataset.groupby(['Pclass'])
group_female_dataset['Survived'].mean()
# higher class survival chance higher
```
